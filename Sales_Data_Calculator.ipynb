# Sales Data Calculator

## Task 4 - AI & ML Internship

This project analyzes a collection of sales values using Python.

The program calculates:

- Total Sales
- Average Sales
- Highest Sale
- Lowest Sale

```python
sales = [1200, 1500, 1800, 1100, 2200, 2500, 1700, 2000, 1600, 1900]
```

```python
print("Sales Data:")
print(sales)
```

```python
total_sales = sum(sales)

print("Total Sales:", total_sales)
```

```python
average_sales = total_sales / len(sales)

print("Average Sales:", average_sales)
```

```python
highest_sale = max(sales)

print("Highest Sale:", highest_sale)
```

```python
lowest_sale = min(sales)

print("Lowest Sale:", lowest_sale)
```

```python
print("----- Sales Analysis -----")
print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)
```

```python
print("----- Sales Analysis -----")
print(f"Total Sales: ₹{total_sales:,.2f}")
print(f"Average Sales: ₹{average_sales:,.2f}")
print(f"Highest Sale: ₹{highest_sale:,.2f}")
print(f"Lowest Sale: ₹{lowest_sale:,.2f}")
```
